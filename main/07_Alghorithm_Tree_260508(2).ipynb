```python
class TreeNode:
    def __init__(self, data, left, right):
        self.data = data
        self.left = left
        self.right = right
    def set_left(self, left):
        self.left = left
    def set_right(self, right):
        self.left = right
```

```python
my_tree = TreeNode('A', None, None)
new_node = TreeNode('B', None, None)
my_tree.set_left(new_node)
new_node = TreeNode('C', None, None)
my_tree.set_right(new_node)

cur_node = my_tree.left
cur_node.set_left(TreeNode('D', None, None))
cur_node.set_right(TreeNode('E', None, None))

cur_node = my_tree.right
```

```python
class STreenode:
    def __init__(self, key, left, right):
        self.key = key
        self.left = left
        self.right = right
```

```python
# BST 생성
t3 = STreenode(3, None, None)
t12 = STreenode(12, None, None)
t7 = STreenode(7, t3, t12)
t27 = STreenode(27, None, None)
t31 = STreenode(31, t27, None)
t26 = STreenode(26, None, t31)
t18 = STreenode(18, t7, t26)
my_bst = t18
```

```python
def search_bst(n, key):
    if n == None: return None
    if key == n.key: return n
    if key < n.key:
        return search_bst(n.left, key)
    else:
        return search_bst(n.right, key)
    
result = search_bst(my_bst, 32)
if result: print(result.key)
else:
    print('없는 key입니다.')
```

```python
def bst(n, key):
    while n != None:
        mom = n
        if key == n.key:
            return False
        elif key < n.key:
            n = n.left
        else:
            n = n.right
    return mom

def preorder(node):
    print(node.key, end=' ')
    if node.left: preorder(node.left)
    if node.right: preorder(node.right)

turn = int(input())
for i in range(turn):
    input_num = int(input('숫자를 입력하세요 : '))
    result = bst(my_bst, input_num)

    if result is False:
        print('이미 있는 값입니다.')
    elif input_num < result.key:
        result.left = STreeNode(input_num, None, None)
        print(f'{result.key}의 왼쪽에 {input_num}입력되었습니다.')
    else:
        result.right = STreeNode(input_num, None, None)
        print(f'{result.key}의 오른쪽에 {input_num}입력되었습니다.')
    preorder(my_bst)
    print()
    
print('[종료]')
```
